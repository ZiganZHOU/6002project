```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys; sys.path.append('..')
from src.preprocess import load_data, impute_missing, get_features_target, split_and_scale

df = load_data('../data/framingham.csv')
print('Shape before imputation:', df.shape)
df = impute_missing(df)
print('Shape after imputation:', df.shape)
```
