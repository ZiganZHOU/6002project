```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys; sys.path.append('..')

from src.preprocess import load_data, summarize_missing, full_pipeline

sns.set_theme(style='whitegrid')

df_raw = load_data('../data/framingham.csv')
print('Shape:', df_raw.shape)
print('\nMissing values (raw):')
print(summarize_missing(df_raw))

# Leakage-safe pipeline (fit imputer/scaler on train only)
X_train, X_test, y_train, y_test, scaler, imputer, feature_names = full_pipeline('../data/framingham.csv')
print('\nPipeline done (split → fit MICE on train → transform test → fit scaler on train → transform test)')
print('Train size:', X_train.shape)
print('Test size: ', X_test.shape)
print('Positive rate (train):', y_train.mean().round(3))
print('Positive rate (test): ', y_test.mean().round(3))
```
